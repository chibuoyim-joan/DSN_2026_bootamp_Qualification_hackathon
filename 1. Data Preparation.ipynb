#### Filling `product_weight_kg`

```python
combined.groupby("product_code").agg({"product_weight_kg": "mean"}).reset_index()
```

```python
combined["product_weight_kg"] = combined["product_weight_kg"].fillna(

    combined.groupby(["product_code"])["product_weight_kg"].transform("mean")
)
```

```python
combined.isna().sum()
```

```python
#  Filling the remianing 4 data points with overall average of the weight
combined["product_weight_kg"].fillna(combined["product_weight_kg"].mean(), inplace = True)
```

#### Normalizing the `product_category` column
- so all text is in lower case

```python
combined["product_category"] = combined["product_category"].str.lower()
```

### Feature Engineering

```python
combined['price_vs_cat'] = combined['product_price'] / combined.groupby('product_category')['product_price'].transform('median')
combined['price_rank_in_cat'] = combined.groupby('product_category')['product_price'].rank(pct=True)
combined['price_vs_store'] = combined['product_price'] / combined.groupby('store_code')['product_price'].transform('mean')

combined['price_per_kg'] = combined['product_price'] / combined['product_weight_kg']
combined['weight_vs_cat'] = combined['product_weight_kg'] / combined.groupby('product_category')['product_weight_kg'].transform('median')

combined["vis_vs_product"] = combined['shelf_visibility'] / combined['shelf_visibility'].mean()
combined['vis_x_price'] = combined['shelf_visibility'] * combined['product_price']
```

```python
combined.head()
```

### Encoding data

#### Encoding `store_size`
- Because these stores have an order, we will be encoding with categorical numbers where the larger store takes the larger number
  - Large = 30
  - Medium = 20
  - Small = 10

```python
combined["store_size"] = combined["store_size"].map({"Large": 30, "Medium": 20, "Small": 10})
```

#### Encoding `store_location_tier`
- Because these store_location_tier have an order, we will be encoding with categorical numbers where the top tier takes the larger number
  - Tier_3 = 30
  - Tier_2 = 20
  - Tier_1 = 10

```python
combined["store_location_tier"] = combined["store_location_tier"].map({"Tier_3": 30, "Tier_2": 20, "Tier_1": 10})
```

#### Encoding `store_format`
- Because these store_format have an order, we will be encoding with categorical numbers where the bigger store takes the larger number
  - Flagship Hypermarket = 40
  - Superstore = 30
  - Standard Supermarket = 20
  - Corner Shop = 10

```python
combined["store_format"] = combined["store_format"].map({"Flagship Hypermarket":40, "Superstore": 30, "Standard Supermarket": 20, "Corner Shop": 10})
```

```python
combined["fat_content"] = combined["fat_content"].map({"Low Fat":20, "Regular": 10})
```

#### Dropping Columns

```python
combined.drop(["product_code", "product_category", "store_code"], axis =1, inplace = True, errors = "ignore")
```
